```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={'Year':[2015,2016,2017,2018,2019],'Sales':[100,150,200,180,220],'Expenses':[80,100,120,100,130]}
df=pd.DataFrame(data)
```

```python
#line plot
df.plot(x='Year',y='Sales',kind='line')
plt.title('Sales Over the Years')
plt.xlabel('Year')
plt.ylabel('Sales (in units)')
plt.show()
```

```python
#Bar Plot
df.plot(x='Year',y='Sales',kind='bar')
plt.title('Sales Over the Years')
plt.xlabel('Year')
plt.ylabel('Sales (in units)')
plt.show()
```

```python
#Scatter Plot
df.plot(x='Sales',y='Expenses',kind='scatter')
plt.title('Expenses VS Sales')
plt.xlabel('Sales')
plt.ylabel('Expenses (in units)')
plt.show()
```

```python
#Histogram
df['Sales'].plot(kind='hist')
plt.title('Sales Distribution')
plt.xlabel('Sales (in units)')
plt.ylabel('Frequency')
plt.show()
```

4.Annotations

```python
import matplotlib.patches as patches
import matplotlib.lines as lines
```

```python
plt.plot([1,2,3,4],[1,4,2,3],label='Series 1')
plt.plot([1,2,3,4],[4,3,1,3],label='Series 2')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Plot Title')

plt.text(1,1,'Example Text Annotation', fontsize=10, color='red') #TEXT ANNOTATION

#plt.annotate('Reversal',xy=(3,2),xytext=(2,1.5),arrowprops=dict{'facecolor':'black','arrowstyle':'->'}) #ARROW
#plt.annotate(" ",xy=(3,1),xytext=(2.4,1.5),arrowprops=dict{'facecolor':'red','arrowstyle':'->'})
plt.annotate('Reversal', xy=(3,2), xytext=(2,1.5),arrowprops=dict(facecolor='black',arrowstyle='->'))
plt.annotate("", xy=(3,1), xytext=(2.4,1.5),arrowprops=dict(facecolor='red',arrowstyle='->'))


circle=patches.Circle((2,4),0.4,facecolor='blue',alpha=0.3)#CIRCLE
plt.gca().add_patch(circle)#CIRCLE

rect=patches.Rectangle((1,4.5),3,0.7,facecolor='red',alpha=0.3)#RECTANGLE
plt.gca().add_patch(rect)#RECTANGLE

line1= lines.Line2D([2.5,3.5],[1,1], linewidth=2,color='green')#LINE
plt.gca().add_line(line1)
line2=lines.Line2D([1.5,2.5],[4,4],linewidth=2,color='purple')#LINE 2
plt.gca().add_line(line2)
plt.legend()
plt.show
```

creating subplot

```python
x=[1,2,3,4,5]
y1=[10,15,12,17,9]
y2=[5,8,6,12,10]
y3=[7,10,9,13,11]
plt.subplot(2,2,1)
plt.plot(x,y1)
plt.title('Subplot 1')

plt.subplot(2,2,2)
plt.plot(x,y2)
plt.title('Subplot 2')

plt.subplot(2,2,3)
plt.plot(x,y3)
plt.title('Subplot 3')

plt.tight_layout()
plt.show()
```

creating pie chart

```python
categories=['A','B','C','D']
values=[30,15,25,30]
plt.pie(values,labels=categories)
plt.title('Pie Chart')
plt.show()
```

```python
colors=['red','green','blue','yellow']
plt.pie(values,labels=categories,colors=colors)
plt.title('Pie Chart')
plt.show()
```

```python
explode=[0.1,0,0,0]
plt.pie(values,labels=categories,explode=explode)
plt.title('Pie Chart')
plt.show()
```

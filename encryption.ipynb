```python
#function for encryption (pt=plain text, kt=key)
def encryp(pt,k):
  result=""

  for i in range(len(pt)):
    char=pt[i]
    if(char.isupper()):
      result+=chr((ord(char)+k-65)%26+65)
    else:
      result+=chr((ord(char)+k-97)%26+97)
  return result
#plain text
pt="TolaniPolytechnic"
#key
k=4
print("Plain Text: "+pt)
print("Shift Key: "+str(k))
print("Cipher Text: "+encryp(pt,k))
```

```python
def decryp(pt,k):
  result=""

  for i in range(len(pt)):
    char=pt[i]
    if(char.isupper()):
      result+=chr((ord(char)-k-65)%26+65)
    else:
      result+=chr((ord(char)-k-97)%26+97)
  return result

pt="XspermTspcxiglrmg"
k=4
print("Cipher Text: "+pt)
print("Shift Key: "+str(k))
print("Plain Text: "+decryp(pt,k))
```

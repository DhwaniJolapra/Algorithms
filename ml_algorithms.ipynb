```python
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

image = Image.open('/content/img6.jpg').convert("RGB")

min_filter = image.filter(ImageFilter.MinFilter(size=3))
max_filter = image.filter(ImageFilter.MaxFilter(size=3))
median_filter = image.filter(ImageFilter.MedianFilter(size=3))

plt.figure(figsize=(10, 8))

plt.subplot(221)
plt.imshow(image)
plt.title("Original Image")

plt.subplot(222)
plt.imshow(min_filter)
plt.title("Min Filter Image")

plt.subplot(223)
plt.imshow(max_filter)
plt.title("Max Filter Image")

plt.subplot(224)
plt.imshow(median_filter)
plt.title("Median Filtered Image")

plt.tight_layout()
plt.show()
```
